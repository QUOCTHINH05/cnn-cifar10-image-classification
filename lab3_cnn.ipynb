```python
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix
from classifier import CIFARMLPClassifier, CIFARCNNClassifier

# ── Cấu hình ──────────────────────────────────────────────────────────────────
DATA_PATH = "data/cifar10.npz"
MODEL_DIR = "models"
FIGURES_DIR = "figures"
CLASS_NAMES = [
    "airplane", "automobile", "bird", "cat", "deer",
    "dog", "frog", "horse", "ship", "truck"
]

def plot_cm(model_path, model_type, x_test, y_test):
    """Load model, dự đoán và vẽ heatmap Confusion Matrix."""
    # 1. Khởi tạo và Load model
    if model_type == "MLP":
        clf = CIFARMLPClassifier()
    else:
        clf = CIFARCNNClassifier()

    clf.load(model_path)

    # 2. Dự đoán nhãn trên tập test
    print(f"Đang thực hiện dự đoán cho {model_type}...")
    y_pred = np.argmax(clf.model.predict(x_test), axis=1)

    # 3. Tính Confusion Matrix
    cm = confusion_matrix(y_test, y_pred)

    # 4. Vẽ biểu đồ bằng Seaborn (đẹp và trực quan hơn matplotlib thuần)
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES)
    plt.title(f"Confusion Matrix - {model_type}")
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')

    # 5. Lưu ảnh
    os.makedirs(FIGURES_DIR, exist_ok=True)
    save_path = os.path.join(FIGURES_DIR, f"confusion_matrix_{model_type.lower()}.png")
    plt.savefig(save_path, dpi=300)
    plt.show()
    print(f"Đã lưu ảnh tại: {save_path}")

def main():
    # Load dữ liệu test
    data = np.load(DATA_PATH)
    x_test, y_test = data["x_test"], data["y_test"]

    # Vẽ cho MLP
    plot_cm(os.path.join(MODEL_DIR, "cifar_mlp_model.keras"), "MLP", x_test, y_test)

    # Vẽ cho CNN
    plot_cm(os.path.join(MODEL_DIR, "cnn_model.keras"), "CNN", x_test, y_test)

if __name__ == "__main__":
    main()
```
